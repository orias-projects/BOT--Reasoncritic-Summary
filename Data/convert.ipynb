```python
import json
from pathlib import Path

input_file = Path("VNExpress.json")   # file nguồn
output_file = Path("VNExpress_dataset.jsonl")  # output phù hợp HuggingFace

with open(input_file, "r", encoding="utf-8") as f:
    data = json.load(f)

count_total = len(data)
count_kept = 0

with open(output_file, "w", encoding="utf-8") as f:
    for item in data:
        # Bỏ các bài có số từ >500
        if item.get("words", 0) > 500:
            continue

        new_item = {
            "article": item.get("content", "").strip(),
            "summary": item.get("description", "").strip()
        }
        f.write(json.dumps(new_item, ensure_ascii=False) + "\n")
        count_kept += 1

print(f"✅ Done! Dataset saved to: {output_file}")
print(f"📦 Tổng số bài: {count_total}")
print(f"👍 Giữ lại: {count_kept}")
print(f"🗑️ Loại bỏ: {count_total - count_kept} (words > 500)")
```
